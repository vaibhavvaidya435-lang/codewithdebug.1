1614. Maximum Nesting Depth of the Parentheses


```python
class Solution(object):
    def maxDepth(self, s):
        depth = 0
        max_depth = 0

        for ch in s:
            if ch == '(':
                depth += 1
                max_depth = max(max_depth, depth)

            elif ch == ')':
                depth -= 1

        return max_depth

s="(1+(2*3)+((8)/4))+1"
obj=Solution()
print(obj.maxDepth(s))

        
```
