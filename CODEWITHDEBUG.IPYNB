```python
#merge sort
def merge_sort(nums):
    if len(nums) <= 1:
        return nums

    mid = len(nums) // 2

    left = nums[:mid]
    right = nums[mid:]

    left = merge_sort(left)
    right = merge_sort(right)

    return merge(left, right)


def merge(left, right):

    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


nums = [5, 3, 8, 1, 2, 7, 4, 6]

print(merge_sort(nums))
```

```python
#max consecutive sequence 
def longestConsecutive(nums):
    numset = set(nums)
    longest = 0
    for num in numset:
        # Start only if num is the beginning of a sequence
        if num - 1 not in numset:
            count = 1
            while num + count in numset:
                count += 1
            longest = max(longest, count)
    return longest

nums = [1, 2, 98, 4, 99, 100, 32, 101]
print(longestConsecutive(nums))
```

```python
#BINARY SEARCH 
def binary_search(nums, target):
    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            return mid

        elif nums[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1
nums=[22,34,56,78,98,87]
target=78
print(binary_search(nums,target))
```

```python
def rotate(matrix):

    n = len(matrix)

    # Step 1: Transpose
    for i in range(n):
        for j in range(i + 1, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]

    # Step 2: Reverse every row
    for i in range(n):
        matrix[i].reverse()


matrix = [[1,2,3],
          [4,5,6],
          [7,8,9]]

rotate(matrix)

print(matrix)
```

```python
class Solution:
    def threeSum(self, nums):
        nums.sort()
        result = []

        for i in range(len(nums) - 2):

            left = i + 1
            right = len(nums) - 1

            while left < right:

                total = nums[i] + nums[left] + nums[right]

                if total == 0:
                    result.append([nums[i], nums[left], nums[right]])
                    left += 1
                    right -= 1

                elif total < 0:
                    left += 1

                else:
                    right -= 1

        return result

nums=[-1,0,1,-2,3,-4,5]
obj=Solution()
print(obj.threeSum(nums))    
```

```python
class Solution:
    def search(self, nums, target):
        low = 0
        high = len(nums) - 1

        while low <= high:

            mid = (low + high) // 2

            if nums[mid] == target:
                return mid

            # Left half is sorted
            if nums[low] <= nums[mid]:

                if nums[low] <= target < nums[mid]:
                    high = mid - 1
                else:
                    low = mid + 1

            # Right half is sorted
            else:

                if nums[mid] < target <= nums[high]:
                    low = mid + 1
                else:
                    high = mid - 1

        return -1

nums=[1,4,5,6,8,9,10,11,15,20]
target=6
obj=Solution()
print(obj.search(nums,target))
```

```python
#search element in rotated sorted array II

nums = [10,11,11,12,12,13,13,13,1,2,3,4]
target = 11

low = 0
high = len(nums) - 1

while low <= high:
    mid = (low + high) // 2

    if nums[mid] == target:
        print(True)
        break

    # Duplicates: cannot decide which side is sorted
    if nums[low] == nums[mid] == nums[high]:
        low += 1
        high -= 1

    # Left half is sorted
    elif nums[low] <= nums[mid]:
        if nums[low] <= target < nums[mid]:
            high = mid - 1
        else:
            low = mid + 1

    # Right half is sorted
    else:
        if nums[mid] < target <= nums[high]:
            low = mid + 1
        else:
            high = mid - 1

else:
    print(False)
```

```python
class Solution:
    def minEatingSpeed(self, piles, h):
        low = 1
        high = max(piles)

        while low <= high:
            mid = (low + high) // 2

            hours = sum((pile + mid - 1) // mid for pile in piles)

            if hours <= h:
                high = mid - 1
            else:
                low = mid + 1

        return low


piles = [3, 4, 7, 11]
h = 8

obj = Solution()
print(obj.minEatingSpeed(piles, h))
```

```python
def koko(piles,h):

    left = 1
    right = max(piles)

    while left < right:

        speed = left + (right - left)//2

        total = 0
        for pile  in piles:
            total = total + (pile + speed -1)//speed

        if total <= h:

            right = speed

        else:

            left = speed + 1


        

    return right
piles = [3,4,7,11]
h = 8

result = koko(piles,h)
print(result)


```

```python
#largest prefix in string 

def prefix(words):

    if len(words) == 0:
        return ""

    base = words[0]
    result = ""

    for i in range(len(base)):

        for word in words[1:]:
            if i == len(word) or word[i] != base[i]:
                return result

        result += base[i]

    return result


words = ["flower", "flight", "flowrs"]

print(prefix(words))
```

```python
s = "vaidyaa"
t = "vaibhav"

mapping_st = {}
mapping_ts = {}

for i in range(len(s)):
    char_st = s[i]
    char_ts = t[i]

    # s -> t
    if char_st in mapping_st:
        if mapping_st[char_st] != char_ts:
            print(False)
            break
    else:
        mapping_st[char_st] = char_ts

    # t -> s
    if char_ts in mapping_ts:
        if mapping_ts[char_ts] != char_st:
            print(False)
            break
    else:
        mapping_ts[char_ts] = char_st

else:
    print(True)
```
