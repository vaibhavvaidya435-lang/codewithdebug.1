```python
#Remove the outermost parenthesis

s="(()())(())"
count=0
result=""

for ch in s:
    if ch=="(":
        count=count+1
        if count>1:
            result=result+ch
    else:
        count=count-1
        if count>0:
            result=result+ch
print(result)                 
```

```python
#REVERSE THE WORD IN STRING

s="The sky is blue"
words=s.split()
words.reverse()
result=" ".join(words)
print(result)
```

```python
#find the largest odd number oin string 

num="1234567890"
n=len(num)

for i in range(n-1,-1,-1):
    if int(num[i])%2==1:
        print(num[0:i+1])
        break
```

```python
def is_rotation(s, t):
    if len(s) != len(t):
        return False

    return t in (s + s)


s = "abcd"
t = "cdab"

print(is_rotation(s, t))
```
